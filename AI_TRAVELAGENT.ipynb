```python
print("===== Travel Agent =====\n")

# Sensor
budget = float(input("Enter your budget (USD): "))
season = input("Enter season (summer/winter/monsoon): ").lower()
trip_type = input("Enter trip type (adventure/relaxation/family/cultural): ").lower()
days = int(input("Enter number of days: "))

print("\n----------- Recommendation -----------")

# Processing
if budget < 300 or days < 2:
    print("Sorry, not enough budget or days for a trip.")

elif trip_type == "adventure" and season == "winter":
    print("Go to Swiss Alps for skiing and trekking.")

elif trip_type == "adventure":
    print("Go to Nepal for trekking and rafting.")

elif trip_type == "relaxation" and budget >= 1500:
    print("Go to Maldives for a luxury beach holiday.")

elif trip_type == "relaxation":
    print("Go to Goa for a budget-friendly beach trip.")

elif trip_type == "family" and days >= 5:
    print("Go to Dubai for family theme parks.")

elif trip_type == "family":
    print("Go for a short local city tour.")

elif trip_type == "cultural":
    print("Go to Turkey for historical and cultural sites.")

else:
    print("Go for a standard city tour package.")

print("---------------------------------------")
```
